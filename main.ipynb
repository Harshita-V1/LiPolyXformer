# PAVIA

```python
%%bash
# convenience variables
MAT="pavia_split.mat"
SCRIPT="LiPolyXformer_pavia.py"
RUNNER="resource_runner_pavia.py"

OUT_BASE="results_pavia"
EPOCHS=20
#LR=0.005

mkdir -p "${OUT_BASE}"

# helper to run a case: args string must be quoted
run_case () {
  NAME="$1"
  ARGS="$2"
  LR="$3"
  OUTDIR="${OUT_BASE}/${NAME}"
  mkdir -p "$OUTDIR"

  # full command you want monitored
  CMD="python3 ${SCRIPT} --mat ${MAT} --epochs ${EPOCHS} ${ARGS} --out_dir ${OUTDIR} --lr ${LR}"


  echo "=============================================="
  echo " Starting run: $NAME"
  echo " Command: $CMD"
  echo " Output dir: ${OUTDIR}"
  echo "----------------------------------------------"

  # call resource_runner.run_with_monitor in-process via a small python snippet
  python3 - <<PY | tee "${OUTDIR}/console.log"
import os, shlex
from resource_runner_pavia import run_with_monitor
cmd = "$CMD"
# timeseries and summary files saved inside the run folder
ts_csv = os.path.join("$OUTDIR", "timeseries.csv")
summary_csv = os.path.join("$OUTDIR", "summary_metrics_gpu.csv")
# run the monitored command as a subprocess (run_in_process=False)
run_with_monitor(cmd, ts_csv=ts_csv, summary_csv=summary_csv, sample_interval=1.0, run_in_process=False)
PY

  echo "Finished run: $NAME"
  echo ""
}

run_case NoPoly "--no_poly" 
run_case NoCross "--no_cross" 
run_case NoSpatial "--attn_spa_off" 
run_case NoSpectral "--attn_spec_off" 
run_case FreezeCoef "--no-coef-learnable" 
run_case NoUncert "--no_uncertainty" 
run_case Proposed "" 

echo "ALL RUNS FINISHED. Results are in ${OUT_BASE}/*"
```
